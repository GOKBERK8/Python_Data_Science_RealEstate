### NA (Eksik) Değerlerin Kaldırılması
Veri setinde bulunan eksik (missing / NA) değerlerin analiz sürecini olumsuz etkilemesini önlemek amacıyla veri setinden çıkarılması işlemi.

### İstatistiksel Analiz
Veri dağılımını, merkezi eğilim ve yayılım ölçülerini incelemek; değişkenler arasındaki ilişkileri değerlendirmek amacıyla tanımlayıcı ve çıkarımsal istatistiksel yöntemlerin uygulanması.

### Metnin Tokenleştirilmesi
Doğal dil işleme (NLP) uygulamalarında kullanılmak üzere metnin kelime, alt kelime veya sembol gibi daha küçük birimlere ayrıştırılması süreci.

### Aykırı Değerlerin Kaldırılması
Veri setinde yer alan ve diğer gözlemlerden anlamlı derecede sapma gösteren aykırı değerlerin, model performansını ve analiz doğruluğunu artırmak amacıyla tespit edilerek veri setinden çıkarılması.

### Boyut İndirgeme
Veri setindeki özellik sayısını, bilgi kaybını en aza indirerek azaltmayı amaçlayan yöntemlerdir. En yaygın kullanılan tekniklerden biri Temel Bileşenler Analizi (Principal Component Analysis – PCA)’dır.

### Özellik Mühendisliği
Mevcut değişkenlerden yeni ve anlamlı özellikler türetme veya mevcut özellikleri dönüştürme yoluyla makine öğrenmesi modellerinin performansını artırmayı hedefleyen süreç.

### Başlangıçtan İleri Seviyeye Veri Görselleştirme
Veri setinin daha iyi anlaşılması amacıyla temel grafiklerden (çubuk grafik, saçılım grafiği vb.) ileri seviye ve etkileşimli görselleştirmelere kadar çeşitli görsel analiz yöntemlerinin kullanılması.

### Grid Search
Bir makine öğrenmesi algoritmasının hiperparametrelerini optimize etmek amacıyla, önceden tanımlanmış parametre uzayı üzerinde kapsamlı arama yapılmasını sağlayan yöntem.

### Lasso Regresyonu
Katsayıları sıfıra doğru küçülten L1 düzenlileştirme (regularization) tekniğini kullanan doğrusal regresyon yöntemidir. Bu sayede değişken seçimi yapılmasına olanak tanır.

### Doğrusal Regresyon
Bir bağımlı değişken ile bir veya daha fazla bağımsız değişken arasındaki doğrusal ilişkiyi modellemeyi amaçlayan istatistiksel ve tahmine dayalı bir yöntem.

### Karar Ağacı Sınıflandırıcısı
Veri setindeki özellikleri kullanarak karar kuralları oluşturan ve bu kurallar aracılığıyla hedef değişkenin sınıfını tahmin eden ağaç tabanlı makine öğrenmesi algoritması.

### Model Değerlendirme
Makine öğrenmesi modellerinin başarımını ölçmek amacıyla doğruluk (accuracy), kesinlik (precision), duyarlılık (recall), F1-skoru ve ROC-AUC gibi performans metriklerinin kullanılması.
